```python
class Human():
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print("大人である")
    else:
      print("大人でない")


human1 = Human("菅田瑛子", 30)
human2 = Human("ちいかわ", 9)
human3 = Human("ハチワレ", 6)

list = [human1, human2, human3]

for human in list:
  human.check_adult()
```
